```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
import os

# === 1. Load the data ===
file_path = r"D:\Research_for_Master\Data_Folder\NRB data for Saroj\For_R\DTF_Events\SPI_SSI Plot\460_SPI_SSI Diff.xlsx"
sheet_name = "Moisture_Data"

df = pd.read_excel(file_path, sheet_name=sheet_name)

# === 2. Rename columns correctly ===
df = df.rename(columns={
    'SWDI_diff': 'Delta_SWDI',
    'P460_SSI_diff': 'Delta_SSI',
    'P904_SPI_diff': 'Delta_SPI',
    'Transition': 'Month_Transition'
})

# Keep only relevant columns
df = df[['Month_Transition', 'Delta_SPI', 'Delta_SSI', 'Delta_SWDI']]

# === 3. Define Event Type (DtF vs Normal) ===
# You can change the threshold based on your definition of DtF
threshold = 1.25
df['Event_Type'] = np.where(df['Delta_SPI'] >= threshold, 'DtF', 'Normal')

# === 4. Function to run PCA ===
def run_pca(sub_df):
    features = ['Delta_SPI', 'Delta_SSI', 'Delta_SWDI']
    X = sub_df[features].dropna()
    if len(X) < 2:  # PCA requires at least 2 samples
        return None
    pca = PCA()
    pca.fit(X)
    return {
        'PC1_var': pca.explained_variance_ratio_[0] * 100,
        'PC2_var': pca.explained_variance_ratio_[1] * 100 if len(pca.explained_variance_ratio_) > 1 else 0,
        'PC1_SPI_loading': pca.components_[0, 0],
        'PC1_SSI_loading': pca.components_[0, 1],
        'PC1_SWDI_loading': pca.components_[0, 2],
        'n_samples': len(X)
    }

# === 5. Apply PCA by Month_Transition × Event_Type ===
records = []
for (month, evt), sub_df in df.groupby(['Month_Transition', 'Event_Type']):
    res = run_pca(sub_df)
    if res:
        res.update({'Month_Transition': month, 'Event_Type': evt})
        records.append(res)

summary_df = pd.DataFrame(records)

# === 6. Order months for seasonal sense ===
month_order = [
    "Jan_Feb","Feb_Mar","Mar_Apr","Apr_May","May_Jun",
    "Jun_Jul","Jul_Aug","Aug_Sep","Sep_Oct","Oct_Nov","Nov_Dec","Dec_Jan"
]
summary_df['Month_Transition'] = pd.Categorical(summary_df['Month_Transition'], categories=month_order, ordered=True)
summary_df = summary_df.sort_values('Month_Transition')

# === 7. Plot PC1 variance across months ===
plt.figure(figsize=(10,5))
sns.lineplot(data=summary_df, x='Month_Transition', y='PC1_var', hue='Event_Type', marker='o')
plt.title("Seasonal Variation of PC1 Variance (ΔSPI–ΔSSI–ΔSWDI Coupling Strength)")
plt.xlabel("Monthly Transition")
plt.ylabel("PC1 Variance Explained (%)")
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# === 8. Plot PCA loadings (SPI, SSI, SWDI) ===
load_melt = summary_df.melt(
    id_vars=['Month_Transition','Event_Type'],
    value_vars=['PC1_SPI_loading','PC1_SSI_loading','PC1_SWDI_loading'],
    var_name='Variable', value_name='Loading'
)

plt.figure(figsize=(12,6))
sns.barplot(data=load_melt, x='Month_Transition', y='Loading', hue='Variable')
plt.title("PC1 Loadings for ΔSPI, ΔSSI, and ΔSWDI per Monthly Transition")
plt.xlabel("Monthly Transition")
plt.ylabel("PC1 Loading Coefficient")
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

# === 9. Save summary ===
output_folder = os.path.dirname(file_path)
output_path = os.path.join(output_folder, "PCA_Results_by_Month_Event.xlsx")
summary_df.to_excel(output_path, index=False)
print(f"PCA summary results saved to: {output_path}")
```
